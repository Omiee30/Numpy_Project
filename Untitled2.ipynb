```python
import numpy as np

np.random.seed(42)

months = np.arange(1, 13)
month_names = np.array([
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
])

stores = np.arange(1, 11)
sales = np.random.randint(10000, 50001, size=(10, 12))

print("STEP 1: SALES DATA")
print("Sales shape:", sales.shape)
print(sales)
print("\nSTEP 2: QUARTERLY SALES")

Q1 = sales[:, 0:3]
Q2 = sales[:, 3:6]
Q3 = sales[:, 6:9]
Q4 = sales[:, 9:12]

print("Q1 Sales:\n", Q1)
print("Q2 Sales:\n", Q2)
print("Q3 Sales:\n", Q3)
print("Q4 Sales:\n", Q4)

print("\nStore 1 monthly sales:")
print(sales[0, :])

print("\nStore 5 Q1 sales:")
print(sales[4, 0:3])
print("\nSTEP 3: SALES FILTERING")
target = 30000
above_target = sales > target

print("\nSales greater than", target, ":")
print(sales[above_target])

store_average = np.mean(sales, axis=1)
overall_average = np.mean(sales)

below_average = store_average < overall_average

print("\nStores below overall average:")
print(stores[below_average])

high_sales = (sales > target) & (sales < 45000)

print("\nSales greater than 30000 AND less than 45000:")
print(sales[high_sales])

special_sales = (sales < 15000) | (sales > 45000)

print("\nSales below 15000 OR above 45000:")
print(sales[special_sales])

not_above_target = ~(sales > target)

print("\nSales less than or equal to target:")
print(sales[not_above_target])

print("\nSTEP 4: FANCY INDEXING")

selected_store_indices = [0, 3, 6]

selected_stores = sales[selected_store_indices, :]

print("\nSales of Stores 1, 4, and 7:")
print(selected_stores)

monthly_totals = np.sum(sales, axis=0)

best_month_indices = np.argsort(monthly_totals)[-3:][::-1]

print("\nBest 3 months:")
for index in best_month_indices:
    print(month_names[index], ":", monthly_totals[index])

print("\nSelected stores' sales in the best 3 months:")
print(sales[np.ix_(selected_store_indices, best_month_indices)])

print("\nSTEP 5: SALES REPORT")

growth_targets = np.linspace(25000, 35000, 12)

print("\nMONTHLY REPORT")
print("----------------------------------")
print("Month   Total Sales   Target   Status")

for i in range(12):
    total = monthly_totals[i]
    target_value = growth_targets[i]

    if total >= target_value:
        status = "Target Achieved"
    else:
        status = "Target Not Achieved"

    print(
        month_names[i],
        "   ",
        total,
        "   ",
        round(target_value, 2),
        "   ",
        status
    )

print("\nSTORE-WISE REPORT")
print("----------------------------------")
print("Store   Total Sales   Average Sales")

store_totals = np.sum(sales, axis=1)

for i in range(10):
    print(
        stores[i],
        "      ",
        store_totals[i],
        "      ",
        round(store_average[i], 2)
    )

print("\nFINAL SUMMARY")
print("----------------------------------")
print("Total yearly sales:", np.sum(sales))
print("Average monthly sales:", round(np.mean(monthly_totals), 2))
print("Best sales month:", month_names[np.argmax(monthly_totals)])
print("Highest store sales: Store", stores[np.argmax(store_totals)])
print("Lowest store sales: Store", stores[np.argmin(store_totals)])
```
